```python
df["Revenue"] = pd.to_numeric(df["Revenue"], errors="coerce")
```

```python
plt.figure(figsize=(10,6))
plt.hist(df["Revenue"], bins=30, edgecolor="black")
plt.title("Distribution of Revenue")
plt.xlabel("Revenue")
plt.ylabel("Number of Orders")
plt.show()
```

Revenue distribution is positively (right) skewed.

What this means:

Most orders have low to medium revenue (roughly ₹100–₹800).
As revenue increases, the number of orders decreases.
Only a few orders generate very high revenue (₹1800–₹2500).

```python
category_revenue = (
    df.groupby("category_name")["Revenue"].sum().sort_values(ascending=False)
)

print(category_revenue)
```

```python
plt.figure(figsize=(10,6))

category_revenue.plot(kind="bar",edgecolor="black", color=sns.color_palette("viridis", len(category_revenue)))

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

Electronic produce the maximum revenue, and fashion generated the least amount of revenue

```python
top_products=(df.groupby("product_name")["Revenue"].sum().sort_values(ascending=False))
print(top_products)
```

```python
plt.figsize=(10,6)
top_products.head(10).plot(kind="bar",edgecolor="black", color=sns.color_palette("viridis", len(top_products.head(10))))
plt.title("Top 10 Products by Revenue")
plt.xlabel("Product Name")
plt.ylabel("Total Revenue") 
```

Smartphones are leading to produce the maximum revenue followed by notebook; and in the top 10 products, T-shirts are generating the lowest amount of revenue.

```python
city_revenue=(df.groupby("city")["Revenue"].sum().sort_values(ascending=False))
print(city_revenue)
```

```python
plt.figsize=(10,6)
city_revenue.head(10).plot(kind="bar",edgecolor="black", color=sns.color_palette("viridis", len(city_revenue.head(10))))
plt.title("Top 10 Cities by Revenue")
plt.xlabel("City")
plt.ylabel("Total Revenue") 
```

```python
df["Age Group"] = pd.cut(df["age"], bins=[0, 18, 25, 35, 45, 55, 65,75], labels=["0-18", "19-25", "26-35", "36-45", "46-55", "56-65", "65+"])
```

```python
age_revenue=(df.groupby("Age Group")["Revenue"].sum())
age_revenue
```

```python
plt.figsize=(10,6)
age_revenue.plot(kind="bar",edgecolor="black", color=sns.color_palette("deep", len(age_revenue)))
plt.title("Revenue by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Total Revenue")
plt.show()
```

```python
review_category=(df.groupby("category_name")["review_score"].mean().sort_values(ascending=False))
```

```python
plt.figsize=(10,6)
review_category.plot(kind="bar",edgecolor="black", color=sns.color_palette("dark", len(review_category)))
plt.title("Average Review Score by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Average Review Score")
plt.show()
```
